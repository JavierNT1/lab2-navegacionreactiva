```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("datos_robot.csv")

df.head()
```

```python
plt.figure(figsize=(10,5))

plt.plot(df["tiempo"],
         df["crudo"],
         label="Sensor crudo")

plt.plot(df["tiempo"],
         df["filtrado"],
         label="Filtro exponencial")

plt.xlabel("Tiempo [s]")
plt.ylabel("Lectura sensor")
plt.title("Sensor frontal: Crudo vs Filtrado")

plt.grid(True)
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(df["tiempo"],
         df["fusion"],
         label="Medición fusionada (z)")

plt.plot(df["tiempo"],
         df["kalman"],
         label="Estimación Kalman")

plt.xlabel("Tiempo [s]")
plt.ylabel("Distancia estimada")
plt.title("Medición fusionada vs Estimación Kalman")

plt.grid(True)
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(df["tiempo"],
         df["distancia"])

plt.xlabel("Tiempo [s]")
plt.ylabel("Distancia [m]")
plt.title("Distancia recorrida por el robot")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(df["tiempo"], df["crudo"] - df["filtrado"])

plt.title("Error entre señal cruda y filtrada")
plt.xlabel("Tiempo [s]")
plt.ylabel("Error")

plt.grid(True)

plt.show()
```

```python
print("Cantidad de muestras:", len(df))

print("Tiempo total:",
      round(df["tiempo"].iloc[-1],2),
      "s")

print("Distancia total:",
      round(df["distancia"].iloc[-1],3),
      "m")
```
